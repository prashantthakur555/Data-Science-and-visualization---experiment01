```python
#26
north = Series({'Jan': 40, 'Feb': 55, 'Mar': 60})
south = Series({'Feb': 20, 'Mar': 25, 'Apr': 30})
total = north + south
print(total)
```

```python
#27
d = {'Red': 12, 'Blue': 18, 'Green': 24}
labels = ['Green', 'Red', 'Yellow', 'Blue']
result = Series(d, index=labels)
print(result)
print(pd.isnull(result))
```

```python
#28
s = Series([90, 75, 82], index=['Ravi', 'Aman', 'Neha'])
d = s.to_dict()
s2 = Series(d)
print(d, "\n", s2)
print("Same data:", s.equals(s2))
```

```python
#29
products = Series([450, 620, 380, 710],
                  index=['Pen', 'Notebook', 'Bag', 'Bottle'])
print(products.values)
print(products.index)
print(products['Bag'])
print(products[products > 500])
print('Pen' in products.index)
print(products.to_dict())
```

```python
#30
data = {'A': 15, 'B': 35, 'C': 25}
s = Series(data, index=['C', 'A', 'D', 'B'])
s.name = 'Observations'
s.index.name = 'Category'
print(s)
print(s[s > 20])
print(pd.notnull(s))
print(s.name, "|", s.index.name)
```
