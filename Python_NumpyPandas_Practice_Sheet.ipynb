Pandas Series

```python
# 11.
import pandas as pd
import numpy as np

books = pd.Series([24, 31, 19, 27],
                  index=["Monday", "Tuesday", "Wednesday", "Thursday"])
print(books)
```

```python
#12
prices = pd.Series([45000, 18000, 22000],
                   index=["Laptop", "Tablet", "Phone"])
print(prices["Tablet"])
```

```python
#13
rainfall = pd.Series([12, 0, 18, 5, 23])
print(rainfall[rainfall > 10])
```

```python
#14
metals = pd.Series({'Gold': 7200, 'Silver': 95, 'Copper': 850})
print(metals.index)
print(metals.values)
```

```python
#15
ratings = pd.Series([3.5, 4.2, 2.8, 4.7])
ratings.name = "Customer Ratings"
ratings.index.name = "Review ID"
print(ratings)
```

```python
#16
s = pd.Series([11, 22, 33], index=['X1', 'X2', 'X3'])
print('X2' in s.index)
print('X4' not in s.index)
```

```python
#17
stock = pd.Series([16, 28, 12], index=['Keyboard', 'Mouse', 'Monitor'])
print(stock.to_dict())
```

```python
#18
marks = pd.Series({'Hindi': 78, 'English': 84, 'Computer': 91},
                  index=['Computer', 'Hindi', 'Art', 'English'])
print(marks)
```

```python
#19
expenses = pd.Series([1200, None, 850, None, 600])
print(expenses.isnull())
print(expenses.notnull())
```

```python
#20
a = pd.Series([8, 15, 22], index=['P', 'Q', 'R'])
b = pd.Series([4, 6, 10], index=['R', 'P', 'Q'])
print(a + b)
```

Section B - Moderate


```python
#21
sales = np.array([
[120, 150, 180],
[140, 160, 175],
[135, 155, 190],
[150, 170, 200]
])
daily_total = sales.sum(axis=1)
print(daily_total)

product_total = sales.sum(axis=0)
print(product_total)

best_day = np.argmax(daily_total) + 1
print("Best day: Day", best_day)
print("Sales on that day:", daily_total.max())
```
